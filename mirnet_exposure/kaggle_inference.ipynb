## 4. Multi-Path Data Resolution


```python
import os, glob
model_key = 'mirnet_exposure'
data_root = '/kaggle/input'
target_dir = f'/kaggle/working/LemGendaryDatasets'
os.makedirs(target_dir, exist_ok=True)

print(f'🔍 [DATA] Resolving manifolds for {model_key}...')
patterns = [f'**/*{model_key.lower()}*', f'**/*{model_key.replace("_", "-")}*', f'**/*{model_key.replace("_", "")}*', '**/lemgendary-*']
found = []
for p in patterns: found.extend(glob.glob(os.path.join(data_root, p), recursive=True))

try:
    struct_cmd = "find /kaggle/input -type d -name 'train' | grep 'images/train'"
    import subprocess
    struct_paths = subprocess.run(struct_cmd, shell=True, capture_output=True, text=True).stdout.strip().split('\n')
    for sp in struct_paths:
        if sp: found.append(os.path.dirname(os.path.dirname(sp)))
except: pass

for d in sorted(list(set(found))):
    if os.path.isdir(d):
        # Handle both lowercase slugs and PascalCase names
        bname = os.path.basename(d)
        links = [bname]
        if bname.lower() != bname: links.append(bname.lower())
        
        for link in links:
            link_name = os.path.join(target_dir, link)
            if not os.path.exists(link_name):
                try: os.symlink(d, link_name)
                except: pass
                print(f'✅ [LINKED] {link} -> {d}')
```

## 6. Stealth Model Loading


```python
import os, base64, torch, glob, shutil
model_key = 'mirnet_exposure'
hub_root = '/kaggle/working/LemGendaryModels'
model_hub_dir = os.path.join(hub_root, model_key)
ckpt_hub_dir = os.path.join(model_hub_dir, 'checkpoints')
os.makedirs(ckpt_hub_dir, exist_ok=True)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# 2026 NUCLEAR: Prioritize Kaggle Inputs and recover to Hub
input_pattern = f'/kaggle/input/**/checkpoints/{model_key}*.pth'
input_ckpts = glob.glob(input_pattern, recursive=True)

if input_ckpts:
    print(f'📡 [RECOVERY] Hydrating hub from Kaggle Inputs...')
    for src in input_ckpts:
        fname = os.path.basename(src)
        dst = os.path.join(ckpt_hub_dir, fname)
        if not os.path.exists(dst) or os.path.getsize(src) > os.path.getsize(dst):
            shutil.copy2(src, dst)
            print(f'   -> [OK] Recovered {fname}')

search_paths = [
    os.path.join(ckpt_hub_dir, f'{model_key}_best.pth'),
    os.path.join(ckpt_hub_dir, f'{model_key}_latest.pth')
]
found = []
for p in search_paths: 
    if os.path.exists(p) and os.path.getsize(p) > 10 * 1024 * 1024: # 10MB Threshold for LFS protection
        found.append(p)

model_path = found[0] if found else None

if model_path:
    print(f'💎 [SOTA] Loading pre-trained weights: {model_path}')
    ckpt = torch.load(model_path, map_location=device, weights_only=False)
    print(f'✅ [OK] Weights anchored on {device}.')
else: print('⚠️ [SOTA] No existing weights found. Starting from scratch.')
```
