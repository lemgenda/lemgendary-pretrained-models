## 4. Multi-Path Data Resolution


```python
import os, glob
model_key = 'mirnet_exposure'
data_root = '/kaggle/input'
target_dir = f'/kaggle/working/LemGendaryDatasets'
os.makedirs(target_dir, exist_ok=True)

print(f'🔍 [DATA] Resolving manifolds for {model_key}...')
patterns = [f'**/*{model_key.lower()}*', f'**/*{model_key.replace("_", "-")}*', f'**/*{model_key.replace("_", "")}*', '**/lemgendary-*']
found = []
for p in patterns: found.extend(glob.glob(os.path.join(data_root, p), recursive=True))

try:
    import subprocess
    struct_cmd = "find /kaggle/input -type d -name 'train' | grep 'images/train'"
    struct_paths = subprocess.run(struct_cmd, shell=True, capture_output=True, text=True).stdout.strip().split('\n')
    for sp in struct_paths:
        if sp: found.append(os.path.dirname(os.path.dirname(sp)))
except: pass

for d in sorted(list(set(found))):
    if os.path.isdir(d):
        bname = os.path.basename(d)
        links = [bname]
        if bname.lower() != bname: links.append(bname.lower())
        for link in links:
            link_name = os.path.join(target_dir, link)
            if not os.path.exists(link_name):
                try: os.symlink(d, link_name)
                except: pass
                print(f'✅ [LINKED] {link} -> {d}')
```

## 5. Checkpoint & Metric Recovery


```python
import os, shutil
model_key = 'mirnet_exposure'
print(f'📡 [RECOVERY] Searching for persistent checkpoints for {model_key}...')
search_target = f'lemgendary_{model_key}_checkpoints'.lower().replace('-', '_')
possible_roots = []
for r, dirs, _ in os.walk('/kaggle/input'):
    for d in dirs:
        if search_target in d.lower().replace('-', '_'):
            possible_roots.append(os.path.join(r, d))
if possible_roots:
    recovery_root = sorted(possible_roots, key=lambda x: x.count(os.sep), reverse=True)[0]
    print(f'   -> [FOUND] Recovery manifold at: {recovery_root}')
    # Sync metrics.csv
    src_m = os.path.join(recovery_root, 'metrics.csv')
    if os.path.exists(src_m):
        shutil.copy2(src_m, '/kaggle/working/lemgendary-training-suite/metrics.csv')
        print('   -> [OK] Recovered metrics.csv')
    # Sync checkpoints
    src_c = os.path.join(recovery_root, 'checkpoints')
    dst_c = '/kaggle/working/lemgendary-training-suite/checkpoints'
    os.makedirs(dst_c, exist_ok=True)
    if os.path.exists(src_c):
        for f in os.listdir(src_c):
            if f.endswith('.pth'):
                shutil.copy2(os.path.join(src_c, f), os.path.join(dst_c, f))
                print(f'   -> [OK] Recovered {f}')
else: print('   -> [SKIP] No persistent checkpoints manifold found.')
```
