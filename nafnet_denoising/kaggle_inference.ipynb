# LemGendary SOTA Inference: NafnetDenoising
This notebook natively executes the explicit LemGendary Neural Architecture topologies directly upon Kaggle cloud hardware for both FP32 PyTorch and ONNX models.

```python
import torch
import onnxruntime as ort
from PIL import Image
import numpy as np
import os

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
```

## 1. PyTorch (FP32) Inference


```python
model_path = '/kaggle/input/lemgendary-nafnet-denoising/LemGendaryNafnetDenoising.pth'
if not os.path.exists(model_path):
    model_path = 'LemGendaryNafnetDenoising.pth' # Local fallback

try:
    model = torch.load(model_path, map_location=device)
    model.eval()
    print("\u2705 PyTorch Model loaded successfully!")
except Exception as e:
    print(f"\u274c Error loading PyTorch model: {e}")
```

## 2. ONNX (FP32) Inference


```python
onnx_path = '/kaggle/input/lemgendary-nafnet-denoising/LemGendaryNafnetDenoising.onnx'
if not os.path.exists(onnx_path):
    onnx_path = 'LemGendaryNafnetDenoising.onnx' # Local fallback

try:
    ort_session = ort.InferenceSession(onnx_path, providers=['CUDAExecutionProvider', 'CPUExecutionProvider'])
    print("\u2705 ONNX Session initialized successfully!")
except Exception as e:
    print(f"\u274c Error initializing ONNX session: {e}")
```
