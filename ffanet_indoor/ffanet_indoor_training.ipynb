## 5. Multi-Path Data Resolution


```python
import os
model_key = 'ffanet_indoor'
data_root = '/kaggle/input'
target_dir = f'/kaggle/working/LemGendaryDatasets'
os.makedirs(target_dir, exist_ok=True)

print(f'⚡ [DATA] Speed-Resolving manifolds for {model_key}...')
found = []
target_slugs = [model_key.lower(), model_key.lower().replace("_", "-"), model_key.lower().replace("_", "")]

# 1. Restricted Breadth-First Scanner (max depth 4, directories only) to bypass FUSE latency
if os.path.exists(data_root):
    try:
        queue = [data_root]
        depths = {data_root: 0}
        while queue:
            curr = queue.pop(0)
            depth = depths[curr]
            if depth > 4: continue
            for item in os.listdir(curr):
                path = os.path.join(curr, item)
                if os.path.isdir(path):
                    item_lower = item.lower()
                    # Prune checkpoint folders and models to speed up dataset search
                    if item_lower in ['models', 'checkpoints', 'weights']:
                        continue
                    depths[path] = depth + 1
                    queue.append(path)
                    
                    is_match = any(slug in item_lower for slug in target_slugs) or 'lemgendary' in item_lower or 'datasets' in item_lower
                    if is_match:
                        # Direct check: does it have images/train?
                        if os.path.exists(os.path.join(path, 'images', 'train')):
                            found.append(path)
                        else:
                            # Nested check (1 level deeper)
                            try:
                                for sub in os.listdir(path):
                                    sub_cand = os.path.join(path, sub)
                                    if os.path.isdir(sub_cand) and os.path.exists(os.path.join(sub_cand, 'images', 'train')):
                                        found.append(sub_cand)
                            except:
                                pass
    except Exception:
        pass

for d in sorted(list(set(found))):
    bname = os.path.basename(d)
    camel_name = "".join([w.capitalize() for w in model_key.split("_")])
    links = [
        bname, bname.lower(),
        f"LemGendized{camel_name}KaggleReady",
        f"LemGendized{camel_name}Large",
        f"{camel_name}KaggleReady",
        f"{camel_name}Large"
    ]
    for link in sorted(list(set(links))):
        link_name = os.path.join(target_dir, link)
        if not os.path.exists(link_name):
            try:
                os.symlink(d, link_name)
                print(f'✅ [LINKED] {link} -> {d}')
            except:
                pass
```

## 6. Checkpoint & Metric Recovery


```python
import os, shutil, glob
model_key = 'ffanet_indoor'
print(f'⚡ [RECOVERY] Speed-Searching for {model_key} checkpoints...')
hub_root = '/kaggle/working/LemGendaryModels'
model_hub_dir = os.path.join(hub_root, model_key)
ckpt_hub_dir = os.path.join(model_hub_dir, 'checkpoints')
os.makedirs(ckpt_hub_dir, exist_ok=True)

try:
    import yaml
    with open('/kaggle/working/lemgendary-training-suite/unified_models_v2.yaml', 'r') as f:
        reg = yaml.safe_load(f)
    reg_filename = reg.get(model_key, {}).get('filename', '')
except: reg_filename = ''

# Identify candidates based on naming to avoid scanning image datasets
target_slugs = [model_key.lower().replace('_', ''), model_key.lower().replace('_', '-'), reg_filename.lower() if reg_filename else '']
target_slugs = [s for s in target_slugs if s]

candidate_dirs = []
data_root = '/kaggle/input'
if os.path.exists(data_root):
    try:
        queue = [data_root]
        depths = {data_root: 0}
        while queue:
            curr = queue.pop(0)
            depth = depths[curr]
            if depth > 4: continue
            for item in os.listdir(curr):
                path = os.path.join(curr, item)
                if os.path.isdir(path):
                    item_lower = item.lower()
                    # Prune image manifolds and datasets directory entirely to bypass FUSE latency
                    if item_lower in ['datasets', 'images', 'train', 'val', 'test', 'validation', 'dataset']:
                        continue
                    depths[path] = depth + 1
                    queue.append(path)
                    
                    if any(slug in item_lower for slug in target_slugs) or 'checkpoint' in item_lower or 'weights' in item_lower or 'models' in item_lower:
                        candidate_dirs.append(path)
    except: pass

found_ckpts = []
for c_dir in candidate_dirs:
    for root, _, files in os.walk(c_dir):
        for f in files:
            if f.endswith('.pth'):
                f_lower = f.lower()
                if any(slug in f_lower for slug in target_slugs) or 'best' in f_lower or 'latest' in f_lower:
                    found_ckpts.append(os.path.join(root, f))

found_ckpts = list(set(found_ckpts))
if found_ckpts:
    print(f'   -> [FOUND] {len(found_ckpts)} binaries in candidate folders.')
    for src in found_ckpts:
        fname = os.path.basename(src)
        target_f = fname
        if 'latest' in fname.lower(): target_f = f'{model_key}_latest.pth'
        elif 'best' in fname.lower(): target_f = f'{model_key}_best.pth'
        elif 'progress' in fname.lower(): target_f = f'{model_key}_progress.pth'
        
        dst = os.path.join(ckpt_hub_dir, target_f)
        if not os.path.exists(dst) or os.path.getsize(src) > os.path.getsize(dst):
            shutil.copy2(src, dst)
            print(f'   -> [OK] Recovered {fname} -> {target_f}')
            
    # Speed-recover metrics.csv
    metrics_found = False
    for src in found_ckpts:
        parent = os.path.dirname(src)
        m_path = os.path.join(parent, 'metrics.csv')
        if os.path.exists(m_path):
            try:
                shutil.copy2(m_path, os.path.join(model_hub_dir, 'metrics.csv'))
                print(f'   -> [OK] Recovered metrics.csv')
                metrics_found = True
                break
            except: pass
else:
    print('   -> [SKIP] No checkpoints found.')
```

## 7. Nuclear Training Matrix


```python
import os, subprocess, sys
os.chdir('/kaggle/working/lemgendary-training-suite')

# 🧹 [JANITOR] Clean up any pre-existing zombie training processes to free the GPU
try:
    current_pid = os.getpid()
    ps_out = subprocess.check_output(['ps', '-ef'], text=True)
    for line in ps_out.split('\n'):
        if 'train.py' in line and str(current_pid) not in line:
            parts = line.split()
            if len(parts) > 1:
                pid = int(parts[1])
                print(f'🧹 [JANITOR] Killing stale zombie training process (PID {pid})...')
                subprocess.run(['kill', '-9', str(pid)], capture_output=True)
except Exception:
    pass

print(f'🚀 [NUCLEAR] Initiating Training Matrix for ffanet_indoor...')
cmd = [sys.executable, '-u', 'training/train.py', '--model', 'ffanet_indoor', '--env', 'kaggle', '--auto_sync']
p = subprocess.Popen(cmd)
try:
    p.wait()
except KeyboardInterrupt:
    print('\n🛑 [TERMINATED] Training interrupted by user. Terminating training subprocess safely...')
    try:
        p.terminate()
        p.wait(timeout=5)
    except subprocess.TimeoutExpired:
        p.kill()
    print('✅ [OK] Subprocess successfully killed. VRAM and CPU are clean.')
```
